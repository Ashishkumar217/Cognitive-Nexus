```python
import json

with open("../outputs/skill_gap.json", "r", encoding="utf-8") as f:
    gap_data = json.load(f)

target_role = gap_data["target_role"]
missing_skills = set(s.lower() for s in gap_data["missing_skills"])

print(f"Target role: {target_role}")
print(f"Missing skills: {missing_skills}")
```

```python
CERTIFICATION_DATABASE = {
    "docker": [{"name": "Docker Certified Associate", "provider": "Docker Inc."}],
    "kubernetes": [{"name": "Certified Kubernetes Application Developer (CKAD)", "provider": "CNCF"}],
    "fastapi": [{"name": "FastAPI Course Certificate", "provider": "Udemy"}],
    "langchain": [{"name": "LangChain for LLM Application Development", "provider": "DeepLearning.AI"}],
    "deep learning": [{"name": "Deep Learning Specialization", "provider": "DeepLearning.AI (Coursera)"}],
    "nlp": [{"name": "Natural Language Processing Specialization", "provider": "DeepLearning.AI (Coursera)"}],
    "machine learning": [{"name": "Machine Learning Specialization", "provider": "Stanford/DeepLearning.AI"}],
    "aws": [{"name": "AWS Certified Cloud Practitioner", "provider": "AWS"}],
    "azure": [{"name": "Microsoft Azure Fundamentals (AZ-900)", "provider": "Microsoft"}],
    "python": [{"name": "Python for Everybody", "provider": "University of Michigan (Coursera)"}],
    "sql": [{"name": "SQL for Data Science", "provider": "IBM (Coursera)"}],
    "data science": [{"name": "IBM Data Science Professional Certificate", "provider": "IBM"}],
    "tensorflow": [{"name": "TensorFlow Developer Certificate", "provider": "Google/TensorFlow"}],
}

GENERAL_CERTS = [
    {"name": "AMD AI PC Developer Certification", "provider": "AMD University Program"},
    {"name": "GitHub Foundations", "provider": "GitHub"}
]

print("Certification database loaded")
```

```python
def recommend_certifications(missing_skills, database=CERTIFICATION_DATABASE, general=GENERAL_CERTS):
    recommended = []
    seen = set()

    for skill in missing_skills:
        if skill in database:
            for cert in database[skill]:
                cert_key = cert["name"]
                if cert_key not in seen:
                    recommended.append({**cert, "for_skill": skill})
                    seen.add(cert_key)

    return {
        "skill_based_certifications": recommended,
        "general_recommendations": general
    }

cert_result = recommend_certifications(missing_skills)

print("Certification Agent: SUCCESS\n")
print("--- SKILL-BASED CERTIFICATIONS ---")
for cert in cert_result["skill_based_certifications"]:
    print(f"• {cert['name']} ({cert['provider']}) — for: {cert['for_skill']}")

print("\n--- GENERAL RECOMMENDATIONS ---")
for cert in cert_result["general_recommendations"]:
    print(f"• {cert['name']} ({cert['provider']})")
```

```python
with open("../outputs/certifications.json", "w", encoding="utf-8") as f:
    json.dump(cert_result, f, indent=2)

print("\nCertifications saved to ../outputs/certifications.json")
print("Notebook 7 (Certification Agent) — COMPLETE")
```
