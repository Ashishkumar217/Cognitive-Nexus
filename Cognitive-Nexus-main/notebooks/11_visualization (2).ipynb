```python
import json
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use('Agg')

def load_json(path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)

skills_data = load_json("../outputs/skills.json")
readiness_data = load_json("../outputs/career_readiness.json")
gap_data = load_json("../outputs/skill_gap.json")
roadmap_data = load_json("../outputs/roadmap.json")
technical_skills_list = skills_data.get("llm_based", {}).get("technical_skills", skills_data["regex_based"]["skills"]["technical"])
soft_skills_list = skills_data.get("llm_based", {}).get("soft_skills", skills_data["regex_based"]["skills"]["soft"])

print("All data loaded for visualization")
```

```python
breakdown = readiness_data["readiness"]["breakdown"]
categories = list(breakdown.keys())
values = list(breakdown.values())

colors = ['#2ecc71' if v >= 50 else '#e74c3c' for v in values]

fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar([c.title() for c in categories], values, color=colors)

for bar, val in zip(bars, values):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 2, f'{val}', ha='center', fontweight='bold')

ax.set_ylim(0, 105)
ax.set_ylabel('Score')
ax.set_title(f"Career Readiness Breakdown — Overall: {readiness_data['readiness']['career_readiness_score']}/100", fontsize=13, fontweight='bold')
ax.axhline(y=50, color='gray', linestyle='--', alpha=0.5, label='Target threshold')
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig("../outputs/chart_readiness_breakdown.png", dpi=150)
plt.show()

print("Chart saved: chart_readiness_breakdown.png")
```

```python
matched_count = len(gap_data["matched_skills"])
missing_count = len(gap_data["missing_skills"])

fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(
    [matched_count, missing_count],
    labels=[f'Matched ({matched_count})', f'Missing ({missing_count})'],
    colors=['#2ecc71', '#e74c3c'],
    autopct='%1.1f%%',
    startangle=90,
    textprops={'fontsize': 11}
)
ax.set_title(f"Skill Match for {gap_data['target_role']}\n({gap_data['match_percentage']}% match)", fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig("../outputs/chart_skill_match.png", dpi=150)
plt.show()

print("Chart saved: chart_skill_match.png")
```

```python
tech_count = len(technical_skills_list)
soft_count = len(soft_skills_list)
fig, ax = plt.subplots(figsize=(6, 5))
bars = ax.bar(['Technical Skills', 'Soft Skills'], [tech_count, soft_count], color=['#3498db', '#9b59b6'])

for bar, val in zip(bars, [tech_count, soft_count]):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.3, f'{val}', ha='center', fontweight='bold', fontsize=12)

ax.set_ylabel('Count')
ax.set_title('Skills Breakdown', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig("../outputs/chart_skills_count.png", dpi=150)
plt.show()

print("Chart saved: chart_skills_count.png")
```

```python
roadmap_items = roadmap_data["roadmap"]

fig, ax = plt.subplots(figsize=(10, 5))

for i, item in enumerate(roadmap_items):
    duration_str = item["duration"].replace("Week ", "")
    if "-" in duration_str:
        start, end = duration_str.split("-")
        weeks = int(end) - int(start) + 1
    else:
        weeks = 1
        start = duration_str

    ax.barh(item["skill"].title(), weeks, left=int(start), color='#1abc9c', edgecolor='black')

ax.set_xlabel('Week')
ax.set_title(f'Learning Roadmap — {roadmap_data["total_duration_weeks"]} Weeks Total', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig("../outputs/chart_roadmap_timeline.png", dpi=150)
plt.show()

print("Chart saved: chart_roadmap_timeline.png")
```
