```python
import json

with open("../outputs/skills.json", "r", encoding="utf-8") as f:
    skills_data = json.load(f)

technical_skills_list = skills_data.get("llm_based", {}).get("technical_skills", skills_data["regex_based"]["skills"]["technical"])
your_skills = set(s.lower() for s in technical_skills_list)

print(f"Loaded {len(your_skills)} technical skills from resume")
print(your_skills)
```

```python
JOB_ROLE_SKILLS = {
    "AI Engineer": [
        "python", "machine learning", "deep learning", "tensorflow", "pytorch",
        "langchain", "docker", "kubernetes", "fastapi", "sql", "rest api",
        "data science", "nlp", "rag"
    ],
    "Data Scientist": [
        "python", "sql", "pandas", "numpy", "machine learning",
        "data science", "tensorflow", "statistics", "data visualization"
    ],
    "Full Stack Developer": [
        "javascript", "html", "css", "react", "node.js", "sql",
        "git", "rest api", "python", "mongodb"
    ],
    "Backend Developer": [
        "python", "java", "sql", "fastapi", "django", "flask",
        "docker", "git", "rest api", "mongodb", "postgresql"
    ]
}

print("Job role skill requirements loaded")
print(f"Available roles: {list(JOB_ROLE_SKILLS.keys())}")
```

```python
def analyze_skill_gap(your_skills, target_role):
    if target_role not in JOB_ROLE_SKILLS:
        return {"error": f"Role '{target_role}' not found. Available: {list(JOB_ROLE_SKILLS.keys())}"}

    required_skills = set(s.lower() for s in JOB_ROLE_SKILLS[target_role])

    matched = your_skills & required_skills
    missing = required_skills - your_skills

    match_percentage = round((len(matched) / len(required_skills)) * 100, 1) if required_skills else 0

    return {
        "target_role": target_role,
        "matched_skills": sorted(list(matched)),
        "missing_skills": sorted(list(missing)),
        "match_percentage": match_percentage
    }

print("Skill Gap Agent defined")
```

```python
target_role = "AI Engineer"

gap_result = analyze_skill_gap(your_skills, target_role)

print("Skill Gap Agent: SUCCESS")
print(f"\nTarget Role: {gap_result['target_role']}")
print(f"Match: {gap_result['match_percentage']}%")
print(f"\nMatched Skills: {gap_result['matched_skills']}")
print(f"\nMissing Skills: {gap_result['missing_skills']}")
```

```python
with open("../outputs/skill_gap.json", "w", encoding="utf-8") as f:
    json.dump(gap_result, f, indent=2)

print("Skill gap data saved to ../outputs/skill_gap.json")
print("Notebook 4 (Skill Gap Agent) — COMPLETE")
```
