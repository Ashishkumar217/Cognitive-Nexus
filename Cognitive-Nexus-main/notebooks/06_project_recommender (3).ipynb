```python
PROJECT_DATABASE = {
    "AI Engineer": {
        "beginner": [
            {"name": "Resume ATS Analyzer", "skills": ["python", "nlp"], "description": "Build a tool that scores resumes against job descriptions"},
            {"name": "Chatbot with Rule-Based Logic", "skills": ["python"], "description": "A simple FAQ chatbot using pattern matching"}
        ],
        "intermediate": [
            {"name": "RAG-based Document Q&A", "skills": ["langchain", "rag", "python"], "description": "Answer questions from PDF documents using retrieval-augmented generation"},
            {"name": "Image Classifier API", "skills": ["tensorflow", "fastapi"], "description": "Deploy a trained CNN model as a REST API"}
        ],
        "advanced": [
            {"name": "Multi-Agent Career Assistant", "skills": ["langchain", "docker", "kubernetes"], "description": "Build a multi-agent system like CareerForge AI itself"},
            {"name": "Fine-tuned LLM Deployment", "skills": ["pytorch", "docker", "fastapi"], "description": "Fine-tune an open-source LLM and deploy it as a scalable API"}
        ]
    },
    "Data Scientist": {
        "beginner": [
            {"name": "Exploratory Data Analysis Dashboard", "skills": ["pandas", "data visualization"], "description": "Analyze a public dataset and visualize insights"}
        ],
        "intermediate": [
            {"name": "Sales Prediction Model", "skills": ["python", "machine learning"], "description": "Predict sales using regression models"}
        ],
        "advanced": [
            {"name": "End-to-End ML Pipeline", "skills": ["python", "docker", "statistics"], "description": "Build a full ML pipeline from data ingestion to deployment"}
        ]
    },
    "Backend Developer": {
        "beginner": [
            {"name": "Todo List REST API", "skills": ["fastapi", "sql"], "description": "A basic CRUD API with database integration"}
        ],
        "intermediate": [
            {"name": "E-commerce Backend", "skills": ["fastapi", "postgresql", "docker"], "description": "Full backend with auth, products, orders"}
        ],
        "advanced": [
            {"name": "Microservices Architecture", "skills": ["docker", "kubernetes", "rest api"], "description": "Split a monolith into communicating microservices"}
        ]
    },
    "Full Stack Developer": {
        "beginner": [
            {"name": "Personal Portfolio Site", "skills": ["html", "css", "javascript"], "description": "A responsive portfolio website"}
        ],
        "intermediate": [
            {"name": "Full Stack Blog Platform", "skills": ["react", "node.js", "mongodb"], "description": "A blog with authentication and CRUD posts"}
        ],
        "advanced": [
            {"name": "Real-time Chat Application", "skills": ["react", "node.js", "rest api"], "description": "Chat app with WebSockets and user presence"}
        ]
    }
}

print("Project database loaded")
```

```python
def recommend_projects(target_role, existing_skills, missing_skills, database=PROJECT_DATABASE):
    if target_role not in database:
        return {"error": f"No projects found for role: {target_role}"}

    recommendations = {"beginner": [], "intermediate": [], "advanced": []}

    for level in ["beginner", "intermediate", "advanced"]:
        for project in database[target_role][level]:
            project_skills = set(project["skills"])
            skill_overlap = project_skills & existing_skills
            skill_needed = project_skills & missing_skills

            recommendations[level].append({
                "name": project["name"],
                "description": project["description"],
                "uses_existing_skills": sorted(list(skill_overlap)),
                "builds_new_skills": sorted(list(skill_needed))
            })

    return recommendations

project_recs = recommend_projects(target_role, existing_skills, missing_skills)

print("Project Recommendation Agent: SUCCESS\n")
for level, projects in project_recs.items():
    print(f"--- {level.upper()} ---")
    for p in projects:
        print(f"• {p['name']}")
        print(f"  {p['description']}")
        print(f"  Uses existing skills: {p['uses_existing_skills']}")
        print(f"  Builds new skills: {p['builds_new_skills']}\n")
```

```python
with open("../outputs/project_recommendations.json", "w", encoding="utf-8") as f:
    json.dump(project_recs, f, indent=2)

print("Project recommendations saved to ../outputs/project_recommendations.json")
print("Notebook 6 (Project Recommendation Agent) — COMPLETE")
```
