```python
import json
from datetime import datetime

def load_json(path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)

skills_data = load_json("../outputs/skills.json")
technical_skills_list = skills_data.get("llm_based", {}).get("technical_skills", skills_data["regex_based"]["skills"]["technical"])
soft_skills_list = skills_data.get("llm_based", {}).get("soft_skills", skills_data["regex_based"]["skills"]["soft"])
education_list = skills_data.get("llm_based", {}).get("education", skills_data["regex_based"]["education"])
projects_list = skills_data.get("llm_based", {}).get("projects", skills_data["regex_based"]["projects"])
readiness_data = load_json("../outputs/career_readiness.json")
gap_data = load_json("../outputs/skill_gap.json")
roadmap_data = load_json("../outputs/roadmap.json")
projects_data = load_json("../outputs/project_recommendations.json")
certs_data = load_json("../outputs/certifications.json")
improvements_data = load_json("../outputs/resume_improvements.json")
interview_data = load_json("../outputs/interview_questions.json")

with open("../outputs/resume_text.txt", "r", encoding="utf-8") as f:
    resume_text = f.read()

print("All 9 previous agent outputs loaded successfully")
```

```python
def generate_final_report():
    report = {
        "generated_on": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
"candidate_summary": {
    "technical_skills_count": len(technical_skills_list),
    "soft_skills_count": len(soft_skills_list),
    "education_entries": len(education_list),
    "projects_count": len(projects_list)
},
        "career_readiness": readiness_data["readiness"],
        "ats_analysis": readiness_data["ats"],
        "skill_gap_analysis": {
            "target_role": gap_data["target_role"],
            "match_percentage": gap_data["match_percentage"],
            "matched_skills": gap_data["matched_skills"],
            "missing_skills": gap_data["missing_skills"]
        },
        "learning_roadmap": {
            "total_weeks": roadmap_data["total_duration_weeks"],
            "plan": roadmap_data["roadmap"]
        },
        "recommended_projects": projects_data,
        "recommended_certifications": certs_data,
        "resume_improvement_suggestions": improvements_data,
        "interview_preparation": interview_data
    }
    return report

final_report = generate_final_report()
print("Final Career Report Generator: SUCCESS")
print("Report structure assembled with all 9 agent outputs")
```

```python
print(f"CANDIDATE SUMMARY")
print(f"  Technical skills: {final_report['candidate_summary']['technical_skills_count']}")
print(f"  Soft skills: {final_report['candidate_summary']['soft_skills_count']}")
print(f"  Education entries: {final_report['candidate_summary']['education_entries']}")
print(f"  Projects: {final_report['candidate_summary']['projects_count']}\n")

print(f"CAREER READINESS SCORE: {final_report['career_readiness']['career_readiness_score']}/100")
for k, v in final_report['career_readiness']['breakdown'].items():
    print(f"  {k.title()}: {v}")
print(f"  Weak areas: {', '.join(final_report['career_readiness']['weak_areas'])}\n")

print(f"ATS SCORE: {final_report['ats_analysis']['ats_score']}/100\n")

print(f"TARGET ROLE: {final_report['skill_gap_analysis']['target_role']}")
print(f"  Skill match: {final_report['skill_gap_analysis']['match_percentage']}%")
print(f"  Missing: {', '.join(final_report['skill_gap_analysis']['missing_skills'])}\n")

print(f"LEARNING ROADMAP: {final_report['learning_roadmap']['total_weeks']} weeks total\n")

print(f"RESUME IMPROVEMENTS: {len(final_report['resume_improvement_suggestions']['rule_based_suggestions'])} rule-based + LLM analysis\n")

interview_prep = final_report['interview_preparation']
template_q = interview_prep.get('template_questions', {})
print(f"INTERVIEW PREP: {len(template_q.get('technical_questions', []))} template technical, "
      f"{len(template_q.get('coding_questions', []))} coding, "
      f"{len(template_q.get('hr_questions', []))} HR, "
      f"{len(template_q.get('behavioral_questions', []))} behavioral, "
      f"+ 5 LLM-personalized questions")

print("\n" + "=" * 70)
print("END OF REPORT")
print("=" * 70)
```

```python
with open("../outputs/final_career_report.json", "w", encoding="utf-8") as f:
    json.dump(final_report, f, indent=2)

print("Final report saved to ../outputs/final_career_report.json")
```

```python
def report_to_markdown(report):
    md = f"""# Cognitive Nexus — Career Report

**Generated:** {report['generated_on']}

## Candidate Summary
- Technical Skills: {report['candidate_summary']['technical_skills_count']}
- Soft Skills: {report['candidate_summary']['soft_skills_count']}
- Education Entries: {report['candidate_summary']['education_entries']}
- Projects: {report['candidate_summary']['projects_count']}

## Career Readiness Score: {report['career_readiness']['career_readiness_score']}/100

| Category | Score |
|---|---|
"""
    for k, v in report['career_readiness']['breakdown'].items():
        md += f"| {k.title()} | {v} |\n"

    md += f"\n**ATS Score:** {report['ats_analysis']['ats_score']}/100\n"

    md += f"\n## Target Role: {report['skill_gap_analysis']['target_role']}\n"
    md += f"**Skill Match:** {report['skill_gap_analysis']['match_percentage']}%\n\n"
    md += f"**Missing Skills:** {', '.join(report['skill_gap_analysis']['missing_skills'])}\n"

    md += f"\n## Learning Roadmap ({report['learning_roadmap']['total_weeks']} weeks)\n"
    for item in report['learning_roadmap']['plan']:
        md += f"- **{item['duration']}**: {item['skill'].title()} — {item['resource']}\n"

    md += f"\n## Resume Improvement Suggestions\n\n"
    md += "### Rule-Based Suggestions\n"
    for s in report['resume_improvement_suggestions']['rule_based_suggestions']:
        md += f"- **[{s['category']}]** {s['suggestion']}\n"

    md += "\n### AI-Generated Deep Analysis\n"
    md += report['resume_improvement_suggestions']['llm_generated_feedback'] + "\n"

    md += f"\n## Interview Preparation\n\n"
    md += "### AI-Personalized Interview Questions\n"
    md += report['interview_preparation'].get('llm_personalized_questions', 'N/A') + "\n"

    return md

markdown_report = report_to_markdown(final_report)

with open("../outputs/final_career_report.md", "w", encoding="utf-8") as f:
    f.write(markdown_report)

print("Markdown report saved to ../outputs/final_career_report.md")
print("\nNotebook 10 (Final Career Report Generator) — COMPLETE")
print("\n" + "=" * 70)
print("ALL 10 NOTEBOOKS COMPLETE — FULL PIPELINE WORKING END TO END")
print("=" * 70)
```
