```python
import json

# Static local dataset simulating job postings - fully offline, no external API
JOB_POSTINGS = [
    {"title": "Junior AI Engineer", "company": "TechNova Solutions", "location": "Bangalore, India",
     "required_skills": ["python", "machine learning", "tensorflow", "sql"], "experience": "0-1 years"},
    {"title": "ML Intern", "company": "DataForge Labs", "location": "Remote",
     "required_skills": ["python", "pandas", "numpy", "machine learning"], "experience": "0 years"},
    {"title": "AI Research Assistant", "company": "Cognitive Systems Inc", "location": "Hyderabad, India",
     "required_skills": ["python", "pytorch", "deep learning", "nlp"], "experience": "0-1 years"},
    {"title": "Backend Developer (Python)", "company": "CloudBridge Technologies", "location": "Pune, India",
     "required_skills": ["python", "flask", "sql", "git"], "experience": "0-2 years"},
    {"title": "Full Stack Developer Trainee", "company": "WebCraft Studio", "location": "Remote",
     "required_skills": ["javascript", "html", "css", "python"], "experience": "0 years"},
    {"title": "Data Science Intern", "company": "Insight Analytics", "location": "Gurugram, India",
     "required_skills": ["python", "pandas", "sql", "machine learning"], "experience": "0 years"},
    {"title": "MLOps Engineer (Entry Level)", "company": "DeployAI", "location": "Bangalore, India",
     "required_skills": ["docker", "kubernetes", "python", "fastapi"], "experience": "0-1 years"},
    {"title": "AI Product Engineer", "company": "NextGen Robotics", "location": "Chennai, India",
     "required_skills": ["python", "tensorflow", "pytorch", "docker"], "experience": "1-2 years"},
]

print(f"Loaded {len(JOB_POSTINGS)} sample job postings (offline local dataset)")
```

```python
with open("../outputs/skills.json", "r", encoding="utf-8") as f:
    skills_data = json.load(f)

your_skills = set(s.lower() for s in skills_data.get("llm_based", {}).get(
    "technical_skills", skills_data["regex_based"]["skills"]["technical"]))

print(f"Your skills: {your_skills}")
```

```python
def match_jobs(your_skills, job_postings, top_n=5):
    matches = []
    for job in job_postings:
        required = set(s.lower() for s in job["required_skills"])
        matched_skills = your_skills & required
        match_pct = round((len(matched_skills) / len(required)) * 100, 1) if required else 0

        matches.append({
            "title": job["title"],
            "company": job["company"],
            "location": job["location"],
            "experience": job["experience"],
            "match_percentage": match_pct,
            "matched_skills": sorted(matched_skills),
            "missing_skills": sorted(required - your_skills)
        })

    matches.sort(key=lambda x: x["match_percentage"], reverse=True)
    return matches[:top_n]

job_matches = match_jobs(your_skills, JOB_POSTINGS)

print("Job Matching Agent: SUCCESS (offline local dataset)\n")
for i, job in enumerate(job_matches, 1):
    print(f"{i}. {job['title']} — {job['company']} ({job['location']})")
    print(f"   Match: {job['match_percentage']}% | Experience: {job['experience']}")
    print(f"   Matched skills: {job['matched_skills']}")
    print(f"   Missing: {job['missing_skills']}\n")
```

```python
with open("../outputs/job_matches.json", "w", encoding="utf-8") as f:
    json.dump(job_matches, f, indent=2)

print("Job matches saved to ../outputs/job_matches.json")
print("Notebook 15 (Offline Job Matcher Agent) — COMPLETE")
print("\nNote: This uses a local sample dataset for demo purposes.")
print("Future scope: integrate with live job board APIs (LinkedIn, Naukri, Indeed) when online access is permitted.")
```
